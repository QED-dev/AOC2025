```python
input_filename = "day1_input.txt"
```

```python
def parse(line):
    if line[0] == 'L':
        return -int(line[1:])
    elif line[0] == 'R':
        return int(line[1:])
    else:
        print(line)
        print(line[0])
        raise
```

```python
def part1(input_filename):
    pos = 50
    password = 0
    with open(input_filename, 'r') as f:
        for line in f:
            pos = (pos + parse(line.strip())) % 100
            if pos == 0:
                password +=1
    return password
part1(input_filename)
```

```python
def part2(input_filename):
    pos = 50
    password = 0
    with open(input_filename, 'r') as f:
        for line in f:
            move = parse(line.strip())
            quotient, newpos = divmod(pos + move, 100)

            if pos == 0 and quotient < 0:
                quotient += 1

            if newpos == 0 and quotient > 0:
                quotient += -1
            
            password += abs(quotient)

            if newpos == 0:
                password +=1

            pos = newpos

    return password
part2(input_filename)
```
