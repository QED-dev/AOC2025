```python
from sympy import divisors
```

```python
input_filename = "day2_input.txt"
```

```python
with open(input_filename, 'r') as f:
    segments = [list(map(int, s.split('-'))) for s in f.read().split(',')]
```

```python
def part1(segment):
    sum = 0
    for id in range(segment[0], segment[1]+1):
        q, r = divmod(len(str(id)),2)
        #print(f"id:{id}, q:{q}, r:{r}, str(id)[:q]:{str(id)[:q]}, str(id)[q:]:{str(id)[q:]}")
        if (r == 0) and (str(id)[:q]==str(id)[q:]):
            sum += id
    return sum


sum([part1(s) for s in segments])
```

```python
def part2(segment):
    sum = 0
    for id in range(segment[0], segment[1]+1):
        id_length = len(str(id))

        for d in divisors(len(str(id)))[:-1]:
            if len(set(str(id)[i:i+d] for i in range(0, id_length, d))) == 1:
                sum += id
                break
            
    return sum


sum([part2(s) for s in segments])
```
