```python
input_filename = "day5_input.txt"
```

```python
def part1(input_filename):
    with open(input_filename) as f:
        input = f.read()
    (ranges, ingredients) = input.split('\n\n')

    ranges = [tuple(map(int,r.split('-'))) for r in ranges.splitlines()]
    ingredients = [int(i) for i in ingredients.splitlines()]

    fresh_count = 0
    for i in ingredients:
        for r in ranges:
            if (r[0]<=i) and (i<=r[1]):
                fresh_count += 1
                break
    return fresh_count

part1(input_filename)
```

```python
def part2(input_filename):

    with open(input_filename) as f:
        input = f.read()
    (ranges, _) = input.split('\n\n')

    ranges = [tuple(map(int,r.split('-'))) for r in ranges.splitlines()]
    ranges = sorted(ranges, key= lambda x: x[0])

    r = ranges[0]
    pos = r[1]
    fresh_count = pos - r[0]+1

    for r in ranges[1:]:
        (a, b) = r

        if pos<a:
            fresh_count += (b-a)+1
            pos=b

        elif (pos>=a) and (pos<=b):
            fresh_count += (b-pos)
            pos=b
    return fresh_count

part2(input_filename)
```
