```python
import math
```

```python
input_filename = "day6_input.txt"
```

```python
def part1(input_filename):
    with open(input_filename, 'r') as f:
        homework = f.read().splitlines()

    homework = [[l for l in line.split(' ') if l != ''] for line in homework]
    operation_row_index = len(homework)-1

    grand_total = 0
    for i in range(len(homework[0])):
        if homework[operation_row_index][i] == '*':
            grand_total += math.prod([int(homework[j][i]) for j in range(operation_row_index)])
        else:
            grand_total += sum([int(homework[j][i]) for j in range(operation_row_index)])
    return grand_total
part1(input_filename)
```

```python
def part2(input_filename):
    with open(input_filename, 'r') as f:
        input = f.read().splitlines()
        numbers = [list(l) for l in input[:-1]]
        symbols = input[-1].strip(' ').split()

    numbers = ["".join([row[i] for row in numbers]).strip() for i in range(len(numbers[0]))]

    problems = [[]]
    for n in numbers:
        if n == "":
            problems.append([])
        else:
            problems[-1].append(n)


    grand_total = 0
    for k in range(len(symbols)):
        problem = [int(x) for x in problems[k]]
        if symbols[k] == '+':
            grand_total += sum(problem)
        else:
            grand_total += math.prod(problem)

    return grand_total
part2(input_filename)
```
