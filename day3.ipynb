```python
input_filename = "day3_input.txt"
```

```python
with open(input_filename, 'r') as f:
    banks = f.read().splitlines()
```

```python
def part1_v1(banks):
    max_voltage = 0
    for bank in banks:
        bank_max_voltage = int(bank[0]+bank[1])
        for i in range(len(bank)):
            for j in range(i+1, len(bank)):
                if int(bank[i] + bank[j])> bank_max_voltage:
                    bank_max_voltage = int(bank[i] + bank[j])
        max_voltage += bank_max_voltage
    return max_voltage

part1_v1(banks)
```

```python
def part1_v2(banks):
    max_voltage = 0
    for bank in banks:
        first_max = max(bank)
        first_max_index = bank.index(first_max)
        if first_max_index == len(bank)-1:
            voltage = max(bank[:-1]) + first_max
        else:
            voltage = first_max + max(bank[first_max_index+1:])
        
        max_voltage += int(voltage)
    return max_voltage
part1_v2(banks)
```

```python
import timeit
iters = 10*4
execution_time = timeit.timeit(lambda: part1_v1(banks), number=iters)
print(f"Execution time version 1: {execution_time} seconds")
execution_time = timeit.timeit(lambda: part1_v2(banks), number=iters)
print(f"Execution time version 2: {execution_time} seconds")
print(f"Execution time version 2 (average): {execution_time/iters} seconds")
```

```python
def recursive_voltage_search(code, bank):
    #print(f"code:{code}, bank:{bank}")

    if len(code) == 12:
        return code
    
    elif len(code + bank) == 12:
        return code + bank
    
    else:
        max_next_max_index = len(bank) - 12 + len(code) + 1

        next_max = max(bank[:max_next_max_index])
        next_max_idx = bank.index(next_max)

        return recursive_voltage_search(code + next_max, bank[next_max_idx+1:])
    
def part2(banks):
    return sum([int(recursive_voltage_search("", bank)) for bank in banks])

part2(banks)
```

```python
import timeit
iters = 10*4
execution_time = timeit.timeit(lambda: part2(banks), number=iters)
print(f"Execution time (average): {execution_time/iters} seconds")
```
